### What to do in this notebook
**Question:** how do we estimate the Kaggle score honestly, before submitting? The validation scheme must reproduce the real situation: we train on the past and predict **later** purchases, the **same customers** appear many times, and the target looks **90 days into the future**.

1. **Describe train → test.** Date range of each set and the gap between them; share of test rows whose customer already appears in train; target rate over time in train.
2. **Why random K-fold is too optimistic.** It trains on the future to predict the past, and puts rows of the same customer on both sides of the split. Keep it only as a reference.
3. **Candidate schemes.**
   - *Time holdout:* sort by `purchase_date`, validate on the latest part of train (similar size to test, ~986 rows).
   - *Expanding-window time CV:* several cut-off dates, always train on the past and validate on the next block.
   - *Group K-fold on `customer_id`:* a customer is never in both train and validation (tests generalisation to unseen customers).
   - *90-day gap:* a training row's target uses purchases up to 90 days later, which may fall inside the validation period. Leaving a 90-day gap between training and validation removes this label leakage.
4. **Compare the schemes** with one simple fixed pipeline (e.g. median imputation + scaling + logistic regression): mean and std of F1 per scheme. Large differences show how optimistic each scheme is.
5. **Check the chosen scheme against test.** Fold sizes, time gap and share of already-seen customers should look like the real train → test split.
6. **Graduate to `src/validation.py`.** A function that returns the splits, plus helpers for out-of-fold predictions and F1 threshold search. Log the choice and the evidence in `docs/decisions.md`.
