### What to do in this notebook
**Question:** what does the data look like, and what is wrong with it? Every finding should lead to a decision (cleaning rule, validation choice, feature idea).

1. **Overview.** Load `train`, `test`, `clientes` from `DATA_RAW`. Check shapes and dtypes and group the columns: current monetary values (`Total a Pagar`, `Total Bruto`, `Total Liquido`, `Total Imp_`, `Total Portes`, `Total Desc_ Global`, `Total Desc_ Linha`, `Custo`), purchase detail (`N_ Detalhes`, `invoice_count_current_purchase`), categoricals (`payment_type`, `salesperson`, `commercial_zone`, `transport_method`, `warehouse`, `document_series`), history (`prior_*`, `days_since_previous_purchase`, `customer_tenure_days`), rolling windows (`purchase_count_*`, `spend_*`), identifiers (`ID`, `customer_id`), `purchase_date`, `random_noise`.
2. **Data quality.** Missing values per column (train vs test). Implausible values: negative totals or costs, cost above the net total, totals that do not add up, rolling counts that are not increasing (30d > 90d is impossible), history values that contradict each other (e.g. `prior_purchase_count = 0` but positive prior spend). Duplicated rows, inconsistent category labels (case, spaces, spelling).
3. **Target balance.** Share of `repeat_purchase_90d = 1`; F1 of a trivial "always 1" prediction as a reference.
4. **Time and customers.** Format and range of `purchase_date` in train vs test; target rate over time; rows per customer; how many test customers already appear in train.
5. **Univariate / multivariate.** Skewness of monetary features, cardinality of categoricals, feature distributions by target class, correlations (the rolling windows are probably redundant), `random_noise` vs target.
6. **`clientes.csv`.** Does every `customer_id` match? Duplicates? Which columns are mostly missing or pure identifiers (street, address, company name)?
7. **Insights.** Numbered list: *finding → action → where it goes* (`02_cleaning`, `03_validation`, `05_features`, `docs/decisions.md`).
